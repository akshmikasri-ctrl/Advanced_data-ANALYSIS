### 1.2 pull up the average isurance charges based on region

```python
insurance_data["region"].unique()
```

```python
insurance_data.groupby("region")["charges"].mean().round(2).sort_values()
```

```python
insurance_data.groupby("region")["charges"].mean().round(2).sort_values(ascending=False)
```

```python
insurance_data.head(7)
```

### 1.3 pull up the average isurance charges based on "region and Sex"
* discrete - Countable/can't be segmented.
* Continuous - measurable/have units/can be segmented..

```python
insurance_data.groupby(["region","sex"])["charges"].mean().round(2).sort_values(ascending=False)
```

## 2.pivot_Table

```python
pd.pivot_table(data=insurance_data,values="charges",index="region",columns="sex") #Presentable comprriing 2 categorical features along with 
                                                                                #1 continuous data.
```

```python
pd.pivot_table(data=insurance_data,values="charges",index="region",columns=["sex","children"])
```

## 3.Aggregate

### pull up the average isurance charges based on region to analyse max,min,sum

```python
insurance_data.groupby("region")["charges"].agg(["mean","min","max","sum"]).round()
```

```python
insurance_data.groupby(["region","sex"])["charges"].agg(["mean","min","max","sum"]).round()
```

## 6.Apply

### create a new column in the name data frame "Eligibility_criteria" where if age is > 25

```python
def get_age(x):
    if x > 25:
        return "Eligible"
    else:
        return "Not Eligibble"
    
```

```python
insurance_data["Eligible Criteria"] = insurance_data["age"].apply(get_age)
insurance_data
```

## 7.Lambda

Anonymous function/nameless function

```python
insurance_data["Eligible Criteria_2"] = insurance_data["age"].apply(lambda x:1 if x>20 else 0)
insurance_data
```

```python
insurance_data["EC"] = insurance_data["age"].apply(lambda x:"Eligible" if x>20 else "Not Eligible")
insurance_data
```
